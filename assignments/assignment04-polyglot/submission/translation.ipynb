```python
import os
import pandas as pd

os.chdir("/workspaces/HDSx-workspace")

nhanes = pd.read_csv("examples/nhanes-equity/data/nhanes_equity_v6.csv")
```

```python
r_summary = (
    nhanes
    .dropna(subset=["BMI", "IncomeGroup", "Gender"])
    .groupby(["IncomeGroup", "Gender"], as_index=False)
    .agg(
        n=("BMI", "size"),
        mean_bmi=("BMI", "mean"),
        sd_bmi=("BMI", "std"),  # pandas uses sample standard deviation by default
    )
    .assign(
        mean_bmi=lambda df: df["mean_bmi"].round(2),
        sd_bmi=lambda df: df["sd_bmi"].round(2),
    )
    .sort_values(["IncomeGroup", "Gender"])
    .reset_index(drop=True)
)

print(r_summary)
```
