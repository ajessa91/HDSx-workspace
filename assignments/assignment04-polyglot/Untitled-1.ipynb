```python
import pandas 
```

```python
nhanes = pandas.read_csv("examples/nhanes-equity/data/nhanes_equity_v6.csv")
nhanes 
```

```python
r_summary = (
    nhanes.dropna(subset=["BMI", "IncomeGroup", "Gender"])
    .groupby(["IncomeGroup", "Gender"], as_index=False)
    .agg(
        n=("BMI", "size"),
        mean_bmi=("BMI", "mean"),
        sd_bmi=("BMI", "std"),  # Sample standard deviation, matching R's sd()
    )
    .round({"mean_bmi": 2, "sd_bmi": 2})
    .sort_values(["IncomeGroup", "Gender"])
    .reset_index(drop=True)
)

r_summary
```
